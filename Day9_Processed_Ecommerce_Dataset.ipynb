# Processed E-commerce Dataset
This notebook demonstrates how to process and combine e-commerce datasets using Python and Pandas.
The datasets used in this assignment are:
- Orders
- Customers
- Products
The main tasks include loading the datasets, inspecting the data, merging related DataFrames, demonstrating concat(), creating useful columns using apply(), performing DateTime operations, organizing the final processed dataset, and exporting it as a CSV file.

```python
import pandas as pd
```

```python
orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")
```

## 1. Inspecting the Datasets
First, we will inspect the Orders, Customers, and Products datasets to understand their structure and contents.

```python
print("Orders Dataset:")
display(orders.head(10))

print("Customers Dataset:")
display(customers.head(10))

print("Products Dataset:")
display(products.head(10))
```

## 2. Merging Orders and Customers
The Orders and Customers datasets are related through the `Customer_ID` column. We will use `merge()` to combine customer information with order details.

```python
orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

display(orders_customers.head(10))
```

## 3. Merging with Products

The merged Orders and Customers DataFrame is now combined with the Products dataset using the `Product_ID` column.

```python
orders_customers_products = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)

display(orders_customers_products.head(10))
```

## 4. Demonstrating concat()

The `concat()` function is used to combine DataFrames along an axis. Here, we demonstrate vertical concatenation by combining two subsets of the processed dataset.

```python
part1 = orders_customers_products.iloc[:5]
part2 = orders_customers_products.iloc[5:10]

concat_demo = pd.concat([part1, part2], ignore_index=True)

display(concat_demo)
```

```python
part1 = orders_customers_products.iloc[:5]
part2 = orders_customers_products.iloc[5:10]
concat_demo = pd.concat([part1, part2], ignore_index=True)
display(concat_demo)
```

## 5. Creating a Total Amount Column Using apply()

The `apply()` function is used to create a new `Total_Amount` column by multiplying the quantity of products ordered by their unit price.

```python
orders_customers_products["Total_Amount"] = orders_customers_products.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

display(orders_customers_products.head(10))
```

## 6. DateTime Operations

The `Order_Date` column is converted to DateTime format so that date-related information can be extracted easily.

```python
orders_customers_products["Order_Date"] = pd.to_datetime(
    orders_customers_products["Order_Date"]
)

display(orders_customers_products[["Order_Date"]].head(10))
```

## 7. Extracting Date Information

After converting the order date to DateTime format, we extract the month, day, and day of the week into separate columns.

```python
orders_customers_products["Order_Month"] = orders_customers_products["Order_Date"].dt.month
orders_customers_products["Order_Day"] = orders_customers_products["Order_Date"].dt.day
orders_customers_products["Order_Day_Name"] = orders_customers_products["Order_Date"].dt.day_name()

display(
    orders_customers_products[
        ["Order_Date", "Order_Month", "Order_Day", "Order_Day_Name"]
    ].head(10)
)
```

## 8. Organizing the Final Processed Dataset

The columns are arranged in a logical order to create a clean and meaningful processed dataset.

```python
final_df = orders_customers_products[
    [
        "Order_ID",
        "Order_Date",
        "Order_Month",
        "Order_Day",
        "Order_Day_Name",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Unit_Price",
        "Quantity",
        "Total_Amount",
        "Payment_Method",
        "Order_Status"
    ]
]

display(final_df.head(10))
```

## 9. Exporting the Processed Dataset

The final processed DataFrame is exported as a CSV file so that it can be used for further analysis and uploaded to GitHub.

```python
final_df.to_csv("Day9_Processed_Ecommerce_Dataset.csv", index=False)

print("Processed dataset exported successfully.")
```

## 10. Conclusion

The Orders, Customers, and Products datasets were successfully loaded and processed using Pandas. The related datasets were combined using `merge()`, and `concat()` was demonstrated for combining DataFrames. A `Total_Amount` column was created using `apply()`. DateTime operations were performed on the order dates to extract the month, day, and day of the week. Finally, the data was organized into a clean processed DataFrame and exported as `Day9_Processed_Ecommerce_Dataset.csv`.
